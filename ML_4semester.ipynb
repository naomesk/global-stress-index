```python
import pandas as pd

df = pd.read_csv("crude-death-rate.csv")

df.info()

df.describe()

print(df.isnull().sum())
```

```python
# I am cleaning my data

# 1. Checking missing values
print(df.isnull().sum())

# 2. Checking duplicated rows
print("Duplicated rows:", df.duplicated().sum())

# 3. Checking data types
print(df.dtypes)

# 4. Checking column names
print(df.columns)

# 5. Checking unique countries
print(df["Entity"].nunique())

# 6. Checking unique country codes
print(df["Code"].nunique())

# 7. Checking the years available
print(df["Year"].min())
print(df["Year"].max())

# 8. Checking unique years
print(sorted(df["Year"].unique()))

# 9. Checking if there are duplicated country/year combinations
print(df.duplicated(subset=["Entity", "Year"]).sum())

# 10. Checking the statistical summary
print(df.describe())
```

```python
# For cleaning the data, I also want to check if there are any outliers.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Charger les données de mortalité pour avoir la bonne colonne
df = pd.read_csv("crude-death-rate.csv")

# Select numerical columns
columns = [
    "Year",
    "Annual crude death rate"
]

# Create boxplots to detect outliers
for col in columns:
    plt.figure(figsize=(8, 3))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.xlabel(col)
    plt.show()
```

```python
# Check the highest death rates

df.nlargest(20, "Annual crude death rate")[
    ["Entity", "Code", "Year", "Annual crude death rate"]
]
```
